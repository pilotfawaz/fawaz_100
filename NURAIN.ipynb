```python
students = ["John", "Mary", "John", "Peter", "Mary", "Grace"]

scores = {
    "John": 85,
    "Mary": 92,
    "Peter": 78,
    "Grace": 95
}

# 1. Remove duplicate student names
unique_students = list(set(students))

# 2. Calculate the total number of unique students
total_students = len(unique_students)

# 3. Calculate the average score
average_score = sum(scores.values()) / total_students

# 4. Find the student with the highest score
top_student = max(scores, key=scores.get)

# Print results
print("2. Unique Students:", total_students)
print("3. Average Score:", average_score)
print("4. Top Student:", top_student.upper())
```
